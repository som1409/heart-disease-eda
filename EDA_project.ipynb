```python
#Create a copy for analysis
df_clean = df.copy()
#fill missing values using the mode
#Mode is suitable here because ca and thal are coded categorical variables.
df_clean["ca"] = df_clean["ca"].fillna(df_clean["ca"].mode()[0])
df_clean["thal"] = df_clean["thal"].fillna(df_clean["thal"].mode()[0])
print("MIssing values after cleaning:")
display(df_clean.isnull().sum())
```

```python
#Create a binary disease-status variable
# target = 0 -> no disease
# target = 1, 2, 3, 4, -> presence of disease
df_clean["disease"] = np.where(
    df_clean["target"] == 0,
    "No Disease",
    "Disease"
)
display(df_clean[["target", "disease"]].head(10))
```

```python
#Group data by target and calculate average values
grouped_analysis = (
    df_clean
    .groupby("target")
    [["age", "trestbps", "chol",
      "thalach", "oldpeak"]]
    .mean()
    .round(2)
)
print("GROUPING AND AGGREGATION")
display(grouped_analysis)
```

```python
#Group by disease status
disease_group = (
    df_clean
    .groupby("disease")
    [["age", "trestbps", "chol",
      "thalach", "oldpeak"]]
    .mean()
    .round(2)
)
print("AVERAGE VALUES BY DISEASE STATUS")
display(disease_group)
```

```python
#Multiple aggregation funtions
aggregation = (
    df_clean
    .groupby("disease")
    .agg({
        "age": ["mean", "min", "max"],
        "chol": ["mean", "min", "max"],
        "thalach": ["mean", "min", "max"]
    })
    .round(2)
)
display(aggregation)
```

```python
#Calculate correlation matrix
numeric_df = df_clean[
    [
        "age",
        "sex",
        "cp",
        "trestbps",
        "chol",
        "fbs",
        "restecg",
        "thalach",
        "exang",
        "oldpeak",
        "slope",
        "ca",
        "thal",
        "target"
    ]
]
correlation_matrix = numeric_df.corr()
print("CORRELATION MAREIX")
display(correlation_matrix.round(2))
```

```python
#Visualize correlation using heatmap
plt.figure(figsize=(12, 9))
sns.heatmap(
    correlation_matrix,
    annot = True,
    fmt = ".2f",
    cmap = "coolwarm",
    linewidths = 0.5
)
plt.title("Correlation Heatmao of Heart Disease Variables")
plt.show()
```

```python
#Correlation of each variable with target
target_correlation = (
    correlation_matrix["target"]
    .sort_values(ascending = False)
)
print("CORRELATION WITH TARGET")
display(target_correlation.round(3).to_frame(name="Correalation"))
```

```python
#Detect potential outliers using IQR
numeric_columns = [
    "age",
    "trestbps",
    "chol",
    "thalach",
    "oldpeak"
]
outlier_summary = []
for column in numeric_columns:
  Q1 = df_clean[column].quantile(0.25)
  Q3 = df_clean[column].quantile(0.75)
  IQR = Q3 - Q1
  lower_limit = Q1 - 1.5 * IQR
  upper_limit = Q3 + 1.5 * IQR
  outliers = df_clean[
      (df_clean[column] < lower_limit) | (df_clean[column] > upper_limit)
  ]
  outlier_summary.append({
      "Variable": column,
      "Q1": round(Q1, 2),
      "Q3": round (Q3, 2),
      "IQR": round(IQR, 2),
      "Lower Limit": round(lower_limit, 2),
      "Upper Limit": round(upper_limit, 2),
      "Potential Outliers": len(outliers)
  })
  outlier_table = pd.DataFrame(outlier_summary)
  display(outlier_table)
```

```python
#Final summary of important variables
final_summary = pd.DataFrame({
    "Measure": [
        "Total Patients",
        "Total Variables",
        "Average Age",
        "Average Cholesterol",
        "Average Maximum Heart Rate",
        "Patients Without Disease",
        "Patients With Disease"
    ],
    "Value": [
        len(df_clean),
        len(columns),
        round(df_clean["age"].mean(), 2),
        round(df_clean["chol"].mean(), 2),
        round(df_clean["thalach"].mean(), 2),
        (df_clean["disease"] == "No Disease").sum(),
        (df_clean["disease"] == "Disease").sum()
    ]
})
display(final_summary)
```
